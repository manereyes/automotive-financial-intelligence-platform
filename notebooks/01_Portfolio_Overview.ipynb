```python
# =====================================================
# IMPORTANDO LIBRERIAS
# =====================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# CONFIGURACION DE DISPLAY Y GRAFICAS
# =====================================================

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
plt.rcParams["figure.figsize"] = (10,6)
```

## Data Acquisition

```python
# Importar la librería de Supabase y dotenv para manejar variables de entorno
from supabase import create_client
from dotenv import load_dotenv

import os
```

```python
# Cargar variables de entorno desde el archivo .env
load_dotenv()
```

### Conexión a Database

```python
# Leer las variables de entorno
SUPABASE_URL = os.getenv("SUPABASE_URL")
SUPABASE_KEY = os.getenv("SUPABASE_KEY")
```

```python
# Creando el cliente de Supabase
supabase = create_client(
    SUPABASE_URL,
    SUPABASE_KEY
)
```

```python
# Validación de la creación del cliente de Supabase
if SUPABASE_URL and SUPABASE_KEY:
    print("✅ Supabase client created successfully.")
else:
    print("❌ Failed to load environment variables.")
```

### Extracción de datos desde Supabase

```python
# Consulta SQL Customers

customers_response = (
    supabase
    .table("customers")
    .select("*")
    .execute()
)

customers_df = pd.DataFrame(customers_response.data)
```

```python
# Consulta SQL Vehicles

vehicles_response = (
    supabase
    .table("vehicles")
    .select("*")
    .execute()
)

vehicles_df = pd.DataFrame(vehicles_response.data)
```

```python
# Consulta SQL Loans

loans_response = (
    supabase
    .table("loans")
    .select("*")
    .execute()
)

loans_df = pd.DataFrame(loans_response.data)
```

- customers_df = Clientes.
- vehicles_df = Inventario financiable.
- loands_df = Operaciones financieras

```python
# Función para hacer preview futuro a DataFrames
def preview_dataframe(df, name):
    """
    Muestra información básica sobre el dataframe dado.

    Parameters
    ----------
    df : pandas.DataFrame
        DataFrame a inspeccionar.
    name : str
        Nombre amigable utilizado en la salida.
    """
    
    print("=" * 60)
    print(f"{name}")
    print("=" * 60)
    print(f"Rows    : {df.shape[0]:,}")
    print(f"Columns : {df.shape[1]}")
    
    display(df.head())
```

```python
# Prueba de función preview_dataframe
preview_dataframe(customers_df, "Customers")
preview_dataframe(vehicles_df, "Vehicles")
preview_dataframe(loans_df, "Loans")
```

## Data Profiling

### Estructura del DataFrame

Objetivo:
- 10,000 filas
- 10 columnas
- Sin duplicados
- 2% valores nulos
- Variables numéricas
- Variables categóricas
- Calidad aceptable

```python
# Función para mostrar información de columnas y tipos de datos
def profile_structure(df, name):
    """
    Muestra la información de estructura del DataFrame, incluyendo el número de filas, columnas y tipos de datos.
    """

    print("=" * 80)
    print(name)
    print("=" * 80)

    print(f"Rows    : {df.shape[0]:,}")
    print(f"Columns : {df.shape[1]}")

    print("\nData Types")

    display(df.dtypes)

    print("\nDataFrame Info")

    df.info()
```

```python
# Prueba de función profile_structure
profile_structure(customers_df, "Customers")
profile_structure(vehicles_df, "Vehicles")
profile_structure(loans_df, "Loans")
```
